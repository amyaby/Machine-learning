## Objective
We want to cluster Pokémon based on their Attack and Defense statistics, but only for Pokémon of types Grass, Fire, or Water. This helps us understand if there are natural groupings among these Pokémon based on their combat stats.


Step 1: Importing Libraries and Loading Data

```python
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/lgreski/pokemonData/refs/heads/master/Pokemon.csv')
df.head(10)
```

Step 2: Filtering and Preparing Data


We are only interested in Pokémon of types Grass, Fire, or Water. This helps us focus our analysis on a specific subset of the data.  
  
We drop the columns Type1, Type2, and Generation because they are not needed for clustering. We only need the Attack and Defense statistics.  
  
We drop non-numerical columns because they are not useful for clustering. We only need the Attack and Defense statistics.  


```python
types = df[['Type1', 'Type2']].isin(['Grass', 'Fire', 'Water'])
drop_cols = ['Type1', 'Type2', 'Generation']
df = df[types].drop(drop_cols, axis=1)
```

Step 3: Applying K-Means Clustering

```python
from sklearn.cluster import KMeans
import numpy as np

kmeans = KMeans(n_clusters=3, random_state=42)
df['cluster'] = kmeans.fit_predict(df[['Attack', 'Defense']])
centroids = kmeans.cluster_centers_
cen_x = [i[0] for i in centroids]
cen_y = [i[1] for i in centroids]
```

Step 4: Visualizing the Clusters

```python
import matplotlib.pyplot as plt

colors = ['R', 'G', 'B']
df['c'] = df['cluster'].map({0: colors[0], 1: colors[1], 2: colors[2]})
plt.scatter(df.Attack, df.Defense, c=df.c, alpha=0.6, s=10)
plt.scatter(cen_x, cen_y, marker='*', c=colors, s=200)
plt.xlabel('Attack')
plt.ylabel('Defense')
plt.title('Pokemon Cluster')
plt.show()
```

Step 5: Enhancing the Visualization with a Legend

```python
from matplotlib.lines import Line2D

legend_elements = [Line2D([0], [0], marker='o', color='w', label='Cluster {}'.format(i+1),
                    markerfacecolor=mcolor, markersize=10) for i, mcolor in enumerate(colors)]
legend_elements.extend([Line2D([0], [0], marker='*', color='w', label='Centroid {}'.format(i+1),
                    markerfacecolor=mcolor, markersize=10) for i, mcolor in enumerate(colors)])

plt.legend(handles=legend_elements, loc='upper right', ncol=2)
plt.xlim(0, 200)
plt.ylim(0, 200)
plt.title('Pokemon Cluster with Centroids')
plt.show()
```
