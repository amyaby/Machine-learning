```python
import pandas as pd
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/tanishq21/Mall-Customers/refs/heads/main/Mall_Customers.csv')
```

```python
df.info()
```

```python
#This line selects columns 3 and 4 (0-based index) from the DataFrame df and converts them to a NumPy array using .values.
df1=df.iloc[:,3:5].values
```

Le code shc.linkage(df1, method='ward') va calculer les distances entre les points de df1 et les fusionner en utilisant la méthode de Ward. Ensuite, shc.dendrogram(...) va tracer un dendrogramme montrant comment ces points sont regroupés.

```python
import scipy.cluster.hierarchy as shc
import matplotlib.pyplot as plt
plt.figure(figsize=(10,7))
plt.title("customer dendograms")
dend = shc.dendrogram(shc.linkage(df1,method='ward'))
```

```python
from sklearn.cluster import AgglomerativeClustering
model = AgglomerativeClustering(n_clusters=5, linkage='ward')
labels_ = model.fit_predict(df1)
```

```python
plt.figure(figsize=(10, 7))
plt.scatter(df1[:,0], df1[:,1], c=model.labels_, cmap='rainbow')
plt.show()
```
